```python
import pandas as pd
import yaml
from IPython.display import display

error_log_path = "../.scratch/cleaned-rule-index/inventory/asca-rule-inventory-error.csv"
```

```python
error_log = pd.read_csv(error_log_path)

error_log.head(3)
```

```python
unknown_features = error_log[error_log["failure_class"] == "unknown_feature"]
unknown_groupings = error_log[error_log["failure_class"] == "unknown_grouping"]
unknown_characters = error_log[error_log["failure_class"] == "unknown_character"]
expected_underscores = error_log[error_log["failure_class"] == "expected_underscore"]

display(unknown_features.head(3))
display(unknown_groupings.head(3))
display(unknown_characters.head(3))
display(expected_underscores.head(3))
```

```python
display_cols = ["section_index","rule_id", "error_token", "description"]

grouping_errors = unknown_groupings[display_cols].copy()
grouping_errors['rule'] = grouping_errors['description'].str.split(" | ", regex=False).str[1]
grouping_errors.drop(columns=['description'], inplace=True)

display(grouping_errors.head(3))

grouping_errors.to_csv("../.scratch/cleaned-rule-index/inventory/grouping_errors.csv", index=False)
```

```python
display_cols = ["section_index","rule_id", "error_token", "description"]

character_errors = unknown_characters[display_cols].copy()
character_errors['rule'] = character_errors['description'].str.split(" | ", regex=False).str[1]
character_errors.drop(columns=['description'], inplace=True)

display(character_errors.head(3))

character_errors.to_csv("../.scratch/cleaned-rule-index/inventory/character_errors.csv", index=False)
```

```python
underscore_errors = expected_underscores[display_cols].copy()
underscore_errors['rule'] = underscore_errors['description'].str.split(" | ", regex=False).str[1]
underscore_errors.drop(columns=['description'], inplace=True)

display(underscore_errors.head(30))
```
