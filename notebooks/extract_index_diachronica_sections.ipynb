```python
from lxml import etree
import pandas as pd
```

```python
tree = etree.parse("./data/index_diachronica.xml")
root = tree.getroot()

results = []
rules = []

def extract_rule_text(el):
    if el.tag == "cite":
        return f"# citation: {el.text}"
    elif el.tag == "rule":
        return f"\t{el.text}"
    elif el.tag == "comment":
        return f"\t# {el.text}"
    else:
        return f"\t# tag: {el.tag} - {el.text}"

for child in root:
    name = child.attrib["name"]
    index = child.attrib["index"]
    rule_count = len(child.findall("rule"))
    results.append({
        "index": index,
        "name": name,
        "rule_count": rule_count,
    })

    rule_text = f"@ {index} - {name}"
    if len(child) > 0:
        rule_text = rule_text + "\n"+ "\n".join([extract_rule_text(el) for el in child])
    rules.append([index, name, rule_text])

sections_df = pd.DataFrame(results)

sections_df.to_csv("./data/index_diachronica_sections.csv", index=False)

sections_df.head(3)
```

```python
for i, rule in enumerate(rules):
    index, name, text = rule
    with open(f"./data/rules/asca/{index}.rsca", "w") as f:
        f.write(text)
```
