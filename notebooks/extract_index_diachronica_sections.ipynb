```python
from lxml import etree
import pandas as pd
from conlanger.tools.SoundChangeRule import SoundChangeRule
```

```python
OVERWRITE = True
```

```python
tree = etree.parse("./data/index_diachronica.xml")
root = tree.getroot()

results = []
rules = []
formats = ("asca", "brassica")

for child in root:
    index = child.attrib["index"]
    results.append({
        "index": index,
        "name": child.attrib["name"],
        "rule_count": len(child.findall("rule")),
    })

    rules.append((index, *(str(SoundChangeRule(child, format=format)) for format in formats)))


sections_df = pd.DataFrame(results)

if OVERWRITE:
    sections_df.to_csv("./data/index_diachronica_sections.csv", index=False)

sections_df.head(3)
```

```python
if OVERWRITE:
    for index, asca_rule, brassica_rule in rules:
        with open(f"./data/rules/asca/{index}.rsca", "w") as f:
            f.write(asca_rule)
        # with open(f"./data/rules/brassica/{index}.bsc", "w") as f:
        #     f.write(brassica_rule)
```
